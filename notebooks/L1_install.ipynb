And if you find a function you're interested in, you can put a question mark at the end of it and will show you the docstring. The docstring *should* give a description of the function and tell you what arguments you're allowed to pass to the function. Here's an example:

```python
si.bandpass_filter?
```

This tells us that `bandpass_filter` expects a `recording` and that we can pass `freq_min` (and more!) arguments to that function. Nice!

If you find a bad docstring, consider rewriting it and submitting an update (PR) to the spikeinterface package. This is a simple way to become a contributor! If you're not sure how to do this, you can read the development guide (https://spikeinterface.readthedocs.io/en/stable/development/development.html). If you're having trouble doing this, please submit an Issue to the SpikeInterface github page and one of the maintinaers will help you to contribute :)
